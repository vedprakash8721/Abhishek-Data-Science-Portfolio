# Q34. What are outliers? How can they affect model performance?

Outliers are unusually high or low values compared to other data points.
They can distort statistical measures such as the mean and standard deviation.
Some Machine Learning algorithms may learn incorrect patterns because of outliers, resulting in poor performance.

# Q37. What is the difference between Normalization and Standardization?

Normalization scales feature values to a fixed range, usually 0 to 1.

Standardization transforms feature values so they have a mean of 0 and a standard deviation of 1.
Standard deviation is a measure of how much the data values are spread out from the mean (average).

```python
# Q39. What is Label Encoding? When should it be used?
# Label Encoding is a technique used to convert categorical labels into numerical values by assigning a unique integer to each category.
import pandas as pd
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df=pd.DataFrame({
    't-shirt':['small','medium','large']
})
df['encoded size']=le.fit_transform(df['t-shirt'])
print(df)
```

```python
# Q40. What is One-Hot Encoding? Why is it preferred for many categorical variables?
"""
Machine learning models cannot understand categorical data directly.
One-Hot Encoding converts each category into a separate column. If a value belongs to a category, it is marked as 1; otherwise, it is marked as 0.
"""
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
df=pd.DataFrame({
    't-shirt':['small','medium','large']
})
enc=OneHotEncoder(sparse_output=False)
encoded=enc.fit_transform(df[['t-shirt']])
encoded_df=pd.DataFrame(encoded,columns=enc.get_feature_names_out(['t-shirt']))
result=pd.concat([df,encoded_df],axis=1)
print(result)
```
