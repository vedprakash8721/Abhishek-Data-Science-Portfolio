Difference from Machine Learning (ML):

# Q10. Explain the complete Machine Learning workflow from data collection to deployment

1 -Data Collection: Gather data from databases, APIs, files, sensors, or other sources.
2 -Data Preprocessing: Clean the data by handling missing values, duplicates, outliers, and formatting issues.
3 -Exploratory Data Analysis (EDA): Analyze the data to understand distributions, relationships, and patterns.
4 -Feature Engineering: Create, modify, or select features that improve model performance.
5 -Train-Test Split: Divide the dataset into training and testing sets.
6 -Model Selection: Choose a suitable Machine Learning algorithm for the problem.
7 -Model Training: Train the selected model using the training data.
8 -Model Evaluation: Measure the model's performance using appropriate evaluation metrics.
9 -Hyperparameter Tuning: Adjust model hyperparameters to improve performance.
10 -Model Deployment: Deploy the trained model so it can make predictions on real-world data.
11 -Monitoring and Maintenance: Continuously monitor the deployed model and update or retrain it when necessary.

# Q11. What are features and target variables?

Features are the input variables (independent variables) used by a Machine Learning model to make predictions.
Target Variable is the output variable (dependent variable) that the model is trained to predict.

# Q12. What is a dataset? Differentiate between training, validation, and testing datasets.

Training Dataset: Used to train the model.
Validation Dataset: Used to tune the model and select the best hyperparameters.
Testing Dataset: Used to evaluate the final performance of the trained model on unseen data.

# Q14. Why is feature engineering important in Machine Learning?

Feature engineering is the process of creating, transforming, or selecting features from raw data to improve the performance of a Machine Learning model.
It converts raw data into meaningful features that the model can learn from

# Q15. What is inference in Machine Learning?

Inference in Machine Learning is the process of using a trained model to make predictions on new, unseen data.
Inference does not update or retrain the model; it only generates predictions.

# Q16. What is model training?

Model training is the process of teaching a Machine Learning model by using training data
so that it learns the relationship between input features and the target variable.

# Q19. What is concept drift?

Concept drift is a situation where the relationship between the input features and the target variable changes over time,
causing a Machine Learning model's performance to degrade.

A model is trained on historical data.
Over time, real-world data patterns may change.
As a result, the model's predictions become less accurate.
The model may need to be retrained with newer data to adapt to the changed patterns.
One Simple Example

A spam detection model trained last year may become less accurate because spammers start using new words and techniques. This change in data patterns is called concept drift.

# Q20. What is data drift?

Data drift is a situation where the distribution of the input features changes over time,
while the relationship between the features and the target variable may remain the same.

A house price prediction model was trained on houses mostly priced between ₹20 lakh and ₹50 lakh.
Later, it receives data where most houses are priced between ₹80 lakh and ₹2 crore. This change in the input data distribution is called data drift.

# Q21. Why is Linear Algebra important in Machine Learning?

Linear Algebra is important in Machine Learning because it provides the mathematical foundation for representing data and performing computations using vectors and matrices.
Machine Learning datasets are represented as matrices.
Features and data points are represented as vectors.
In a house price prediction dataset, each house is represented as a feature vector (e.g., area, bedrooms, age), and all houses together form a matrix, which the model uses for training.

# Q25. What are Mean, Median, and Mode? When is each one preferred?

Mean: The arithmetic average of all values.
Median: The middle value after arranging the data in ascending or descending order.
Mode: The value that occurs most frequently in the dataset.

# Q26. What is Variance?

Variance measures the dispersion or spread of data.
A low variance means the data points are close to the mean.
A high variance means the data points are widely spread from the mean.

It is the average of the squared differences between each data point and the mean.
Since the differences are squared, variance is always greater than or equal to zero

# Q27. What is Standard Deviation?

Standard Deviation (SD) is a statistical measure that indicates how much the data points deviate from the mean (average). It is the square root of the variance.

# Q28. What is Covariance?

Covariance is a statistical measure that indicates how two variables change together. It shows whether the variables move in the same direction or in opposite directions.

Detailed Explanation
Positive Covariance (> 0): Both variables increase or decrease together.
Negative Covariance (< 0): One variable increases while the other decreases.
Zero Covariance (≈ 0): There is no linear relationship between the variables.
