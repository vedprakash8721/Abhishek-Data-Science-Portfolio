# Q1. What is Artificial Intelligence (AI)? How does it differ from Machine Learning?

Artificial Intelligence (AI) is the field of computer science that focuses on creating systems capable of performing tasks that normally require human intelligence, such as learning, reasoning, problem-solving, decision-making, and understanding language.

Difference from Machine Learning (ML):
Machine Learning (ML) is a branch of Artificial Intelligence that enables computers to learn patterns from data and make predictions or decisions without being explicitly programmed for every task.

# Q2. What is Machine Learning? How is it different from traditional programming?

Machine Learning (ML) is a branch of Artificial Intelligence that enables computers to learn patterns from data and make predictions or decisions without being explicitly programmed for every task.

Difference from Traditional Programming:
In traditional programming, the programmer writes explicit rules to produce the desired output. In Machine Learning, the model learns the rules automatically from data.

# Q3. What is Deep Learning? How does it differ from Machine Learning?

Deep Learning (DL) is a subset of Machine Learning that uses artificial neural networks with multiple layers to automatically learn complex patterns from large amounts of data.

# Q4. Explain the different types of Machine Learning. 

Supervised Learning is based on labeled data ,i.e The model is trained using input-output pairs and learns to predict the correct output for new data.
example: spam detection,house price prediction,criket score prediction.

Unsupervised Learning is based on unlabeled data,i.e The model identifies hidden patterns, relationships, or groups in data without target labels.
example: customer segmentation , grouping customer based on behavior.

Reinforcement Learning used agent and environment in which The model learns by interacting with an environment and receiving rewards or penalties based on outcomes.
example: self driving cars.

# Q10. Explain the complete Machine Learning workflow from data collection to deployment

1 -Data Collection: Gather data from databases, APIs, files, sensors, or other sources.
2 -Data Preprocessing: Clean the data by handling missing values, duplicates, outliers, and formatting issues.
3 -Exploratory Data Analysis (EDA): Analyze the data to understand distributions, relationships, and patterns.
4 -Feature Engineering: Create, modify, or select features that improve model performance.
5 -Train-Test Split: Divide the dataset into training and testing sets.
6 -Model Selection: Choose a suitable Machine Learning algorithm for the problem.
7 -Model Training: Train the selected model using the training data.
8 -Model Evaluation: Measure the model's performance using appropriate evaluation metrics.
9 -Hyperparameter Tuning: Adjust model hyperparameters to improve performance.
10 -Model Deployment: Deploy the trained model so it can make predictions on real-world data.
11 -Monitoring and Maintenance: Continuously monitor the deployed model and update or retrain it when necessary.
