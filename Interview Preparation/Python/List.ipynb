```python
#Q69. What is the difference between append(), extend(), and insert()? 
"""append() adds one element at the end of a list.
extend() adds multiple elements from an iterable to the end of a list.
insert() adds one element at a specific index."""
# append:
x=[1,2]
x.append(3)
print(x)

# extend:
x.extend([3,4])
print(x)

# insert:
x.insert(2,2)
print(x)
```

```python
#Q70. What is the difference between remove(), pop(), clear(), and del? 
"""remove() removes an element by value.
pop() removes an element by index and returns it.
clear() removes all elements from the list.
del deletes an element, a range of elements, or the entire list."""
list1=[10,20,30,40,50,60,70]
list1.remove(10)
print(list1)
list1.pop(0)
print(list1)
del list1[0]
print(list1)
list1.clear()
print(list1)
```

```python
#Q73. Write a Python program to remove duplicate elements from a list while preserving the original order.
numbers=[1,2,2,2,2,3,4,5,6,7,7,7,7,8,8,8,9,9,5,5,4,5,5,4,5,6,4]
uni=[]
for num in numbers:
    if num not in uni:
        uni.append(num)
print(uni)
```

```python
#Q74. Write a Python program to find the second largest element in a list. 
nums=[10,20,15,25,221,40]
uniq=list(set(nums))
uniq.sort(reverse=True)
print(uniq[1])
```
