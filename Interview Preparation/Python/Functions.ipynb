```python
# Q42. What is the difference between parameters and arguments? 
"""Parameters are the variables defined in a function declaration,
while arguments are the actual values passed to the function when it is called."""

#name is parameter
def greet(name):
    print("Hello",name)
greet("suhani") # suhani is argument
```

```python
#Q43. What is the difference between positional arguments and keyword arguments? 
"""Positional arguments
Values are passed in the same order as the parameters.
The order of arguments is important.
Keyword arguments
Values are passed using the parameter names.
The order of arguments is not important."""

def student(name,age):
    print(name,age)
student("suhani",23) # positional argument
student(name="ved",age=25) # keyword argument
```

```python
#Q44. What are default arguments? What are their advantages? 
"""Default arguments are function parameters that have predefined values.
If no value is passed during the function call, the default value is used."""

def ex(name='default'):
    print("hello",name)
ex() # default would be used
ex("alice")#overrides default argument
```

```python
#Q45. What are variable-length arguments? Explain *args( star args ) and **kwargs( double star keyword args ). 
""" Variable-length arguments allow a function to accept any number of arguments.
*args is used to accept multiple positional arguments, and **kwargs is used to accept multiple keyword arguments."""

# *args:
def ar(*args):
    print(args)
ar(10,20,30)

# **kwargs:
def kw(**kwargs):
    print(kwargs)
kw(name="ved",relation="owner")
```

```python
#Q46. What is recursion? When should it be used? 
"""Recursion is a programming technique in which a function calls itself to solve a problem.
It should be used when a problem can be broken down into smaller, similar subproblems.
A recursive function calls itself until a base case is reached.
The base case stops the recursion and prevents infinite function calls.
"""
def factorial(n):
    if n==0: # base condition
        return 1
    return n*factorial(n-1)
factorial(6)
```

```python
#Q47. What are the advantages and disadvantages of recursion compared to iteration? 
# make problem simple using repeative function call but takes much memory cauase each function call is stored in call stack.
def factorial(n):
    if n==0:
        return 1
    return n*factorial(n-1)
print(factorial(6))

def factorial_iter(n):
    result=1
    for i in range(1,n+1):
        result*=i
    return result
print(factorial_iter(6))
```

```python
# Q53. Write a recursive function to calculate the factorial of a number. 
n=int(input('enter the number to calculate factorial : '))
def fact(n):
    if n==0:
        return 1
    else:
        return n*fact(n-1)
print(fact(n))
```

```python
# Q54. Write a recursive function to generate the Fibonacci sequence.
n=int(input("enter the num to print fibonacci series : "))
def fib(n):
    if n<=1:
        return n
    else:
        return fib(n-1)+fib(n-2)
    
for i in range(n):
    print(fib(i),end=' ')
```

```python
# Q55. Write a function that accepts any number of integers and returns their sum. 
def cal_sum(*args):
    return sum(args)
print(cal_sum(10,18,12))
```

```python
# Q56. Write a function that accepts arbitrary keyword arguments and prints them in a readable format.
def display_info(**kwargs):
    for key,value in kwargs.items():
        print(f"key : {key}, value : {value}")
display_info(name='ved',age=10)
```
