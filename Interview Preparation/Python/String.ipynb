```python
# Q57. Why are strings immutable in Python? 
"""Strings are immutable in Python, meaning once a string is created, its contents cannot be changed.
Any modification creates a new string object instead of changing the existing one."""
name='Python'
print(id(name))
name=name+' programming'
print(name)
print(id(name))
```

```python
# Q58. What is the difference between find() and index()?
"""Both find() and index() are used to find the position of a substring in a string.

find() returns -1 if the substring is not found.
index() raises a ValueError if the substring is not found.
"""
text1="ved love python "
print(text1.find('love'))
print(text1.find('java'))
print(text1.index('python'))
```

```python
#Q59. What is the difference between split() and join()? 
"""split() breaks a string into a list of strings based on a separator.
join() combines multiple strings into a single string using a separator."""
Data_Science='python sql machine_learning'
result=Data_Science.split(" ")
print(result)
result_join=' '.join(result)
print(result_join)
```

```python
#Q61. What is the difference between strip(), lstrip(), and rstrip()? 
"""strip() removes characters from both ends of a string.
lstrip() removes characters from the left side.
rstrip() removes characters from the right side."""

text='    python    '
print(text.strip())
print(text.rstrip())
print(text.lstrip())
```

```python
#Q63. Write a Python program to count the number of vowels and consonants in a string. 
text=input(" Enter the String ")
vowels=0
consonants=0
for char in text.lower():
    if char in 'aeiou':
        vowels+=1
    elif char.isalpha():
            consonants+=1
print(f" no of vowels is : {vowels}")
print(f"no of consonant is : {consonants}")
```

```python
#Q68. Write a Python program to compress a string using character counts (e.g., aaabbc → a3b2c1). 
text=input('enter the string like aaaabbbc')
result=''
count=1
for i in range(len(text)):
    if i+1<len(text) and text[i]==text[i+1]:
        count+=1
    else:
        result+=text[i]+str(count)
        count=1
print(result)
```
