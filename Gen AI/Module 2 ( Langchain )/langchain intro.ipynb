LangChain is an open-source development framework designed to build applications powered by Large Language Models (LLMs). It acts as a bridge between LLMs (like GPT or Claude) and external resources, making it easier to connect models to databases, APIs, and private documents.
Document Q&A (RAG): Chat with your private data (PDFs, Notion pages, databases) by having the LLM retrieve relevant information before answering.

Conversational Chatbots: Build smart chat interfaces equipped with long-term memory and the ability to maintain context over multiple turns

LangChain Components:
1: Schema: a schema is a standardized data structure or blueprint used to define how information enters, flows through, or exits a component. Instead of dealing with raw text strings, schemas ensure that data is cleanly typed and structured.


2: Models: pretrained models provided by langchain and they are called as 1P first party models.
and models from other company or sources are called as 3P models < gpt,cluade,gemini>.


3: Prompts: prompts are text that are use to interact with models.
4: Indexes: Langchain provides various indexes that facilitates rapid data retrieval.
these indexes act as key to fast searching for data.
in simple words when user give pdf to the llm model then model scan the pdf at one time and make the indexes for upcoming question to ans efficiently.

5: Memory: Memory refers to data storage that used by an app. in langchain memory refers to the storage that stores the conversation that is happened before . 
there are two types of memory
a: conversational memory:it stores the conversational data like chat exchanged and llm responses as well.
b: Longterm Memory: long term memory refers to the data or fact that dont change over time.

6: Chains: chains represents sequence of steps that is used to execute the task. and langchain offers some predefined chains that a developer can use.

7: Agents: agents are software program that are responsible for performing task. langchain provide a set of predefined agents.
