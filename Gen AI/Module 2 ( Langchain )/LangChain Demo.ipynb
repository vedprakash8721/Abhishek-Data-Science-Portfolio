```python
from langchain_ollama import ChatOllama

llm=ChatOllama(model='llama3.2:latest')
ques='what is temperature in large language model?'
response=llm.invoke(ques)
print(response.content)
```

```python
from langchain_ollama import ChatOllama
from langchain_core.callbacks import StreamingStdOutCallbackHandler

# Streaming callback add karne se text live print hoga
llm = ChatOllama(model="llama3.2:latest", callbacks=[StreamingStdOutCallbackHandler()])

ques = "what is temperature in large language model?"
response = llm.invoke(ques)
# Yahan print karne ki zaroorat nahi padegi, streaming khud print kar dega
```

```python
from langchain_core.prompts import PromptTemplate
from IPython.display import display
prompt_template_name=PromptTemplate(
    input_variables=['name'],
    template='what is {name}in short?'
)
pm=prompt_template_name.format(name='langchain')
result=llm.invoke(pm)
display(result)
```

```python
# Chains:
prompt=prompt_template_name.from_template('what is use of {tech}in agentic ai in short?')
prompt.format(tech='langchain')
llm=ChatOllama(model='llama3.2:latest')
chain=prompt | llm
response=chain.invoke({'tech':'langchain'})
print(response.content)
```
