What is a Sequential Chain?
A Sequential Chain is a series of multiple steps or LLM calls executed one after another, where the output or result of the first step is automatically passed as an input to the next step.

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
```
