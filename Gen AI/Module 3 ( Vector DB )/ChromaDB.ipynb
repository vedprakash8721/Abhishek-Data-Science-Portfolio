```python
from langchain_community.document_loaders import DirectoryLoader , TextLoader
loader=DirectoryLoader("./",glob='*.txt',loader_cls=TextLoader)
documents=loader.load()
print(documents)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
def splits_docs(documents,chunk_size=100,chunk_overlap=10):
    text_splitter=RecursiveCharacterTextSplitter(chunk_size=chunk_size,chunk_overlap=chunk_overlap)
    docs=text_splitter.split_documents(documents)
    return docs
docs=splits_docs(documents)
print(len(docs))
from langchain_community.embeddings import SentenceTransformerEmbeddings
embeddings=SentenceTransformerEmbeddings(model_name='all-MiniLM-L6-v2')
```

```python
from langchain_chroma import Chroma
db=Chroma.from_documents(docs,embeddings)
```

```python
query='ollama? '
matching_docs=db.similarity_search(query)
print(matching_docs[0])
```

```python
# printing with matching score
matching_score=db.similarity_search_with_score(query,k=2)
from IPython.display import display
display(matching_score)
```

```python
persist_dir='chroma_db'
vectordb=Chroma.from_documents(
    documents=docs,embedding=embeddings,persist_directory=persist_dir
)
print("data successfully stored in chroma_db")
```

```python
# now using that saved database for search
new_db=Chroma(persist_directory=persist_dir,embedding_function=embeddings)
matching_docs=new_db.similarity_search(query)
print(matching_docs[0])
```
