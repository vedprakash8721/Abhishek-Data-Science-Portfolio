```python
import requests
API_URL = "https://router.huggingface.co/v1/chat/completions"
key='[REDACTED]'
```

```python
def query(payload):
    response=requests.post(
        API_URL,headers={'authorization':f"Bearer {key}"},json=payload
    )
    print(response)
    return response.json()
payload={
    'model': 'google/gemma-7b',
    'messages':[
        {
            'role':'user',
        'content':'what is hugging face in one line?'
        }
    ],'max_tokens':50
}
output=query(payload)
print(output)
```

```python
from transformers import pipeline

model_path = r"C:\MY SPACE\Data Science and Gen AI\Qwen2.5-1.5B-Instruct"

generator = pipeline(
    task="text-generation",
    model=model_path,
    tokenizer=model_path,
    device="cpu"
)
print("Model loaded successfully!")
```

```python
prompt='what is hugging face ? '
result=generator(prompt,
                 temperature=0.7,
                 max_new_tokens=100,
                 do_sample=True)
print(result)
```
