```python
from transformers import pipeline
generator=pipeline(
    task='text-generation',
    model='Qwen/Qwen2.5-1.5B-Instruct'
)
```

```python
prompt='what is hugging face ? '
result=generator(prompt,
                 temperature=0.7,
                 max_new_tokens=100,
                 do_sample=True)
print(result[0]['generated_text'])
```

```python
!pip install -U "bitsandbytes>=0.46.1"
prompts=[
    'what is langchain vs langraph',
    'what is use of langchain and langraph in llm and agentic ai?'
]
from transformers import pipeline
generator=pipeline(
    task='text-generation',
    model='antareepdey/LlamaCare-3B-4Bit',
    device_map='auto'
)
```

```python
for prompt in prompts:
  result=generator(prompt,
                   max_new_tokens=50,
                   temperature=0.7,
                   do_sample=True
                   )
  print(result[0]['generated_text'])
  print('\n')
```
