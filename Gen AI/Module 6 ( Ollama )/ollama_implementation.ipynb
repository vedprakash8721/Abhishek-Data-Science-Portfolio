```python
# Chat with Ollama:
import ollama
messages=[{
    'role': 'system',
    'content':'you are my personal ai assistant'
}]
while True:
    question=input('enter what you want and to end the chat type exit ')
    if question=='exit':
        print('thankyou for the conversation , bye ')
        break
    messages.append({
        'role':'user',
        'content':question
    })
    response=ollama.chat(
        model='qwen2.5:3b',
        messages=messages
    )
    answer=response['message']['content']
    print('AI response:',answer)
    messages.append({
        'role':'assistant',
        'content':answer
    })
```

```python
from transformers import pipeline

# Load model directly onto Colab's GPU
pipe = pipeline(
    'text-generation', model='Qwen/Qwen2.5-Coder-0.5B-Instruct', device='cuda'
)

messages = [{'role': 'system', 'content': 'you are my personal ai assistant'}]

print("AI assistant ready! Type 'exit' to quit.")
while True:
  question = input('\nEnter your question: ')
  if question.strip().lower() == 'exit':
    print('thankyou for the conversation , bye ')
    break

  messages.append({'role': 'user', 'content': question})

  # Generate response using GPU pipeline
  outputs = pipe(messages, max_new_tokens=256)
  answer = outputs[-1]['generated_text'][-1]['content']

  print('AI response:', answer)
  messages.append({'role': 'assistant', 'content': answer})
```
