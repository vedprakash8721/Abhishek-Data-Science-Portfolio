# Modelfile – Customize an LLM
A Modelfile is a configuration file used by Ollama to create a customized model based on an existing model.

For example, suppose you have Qwen 2.5 3B, but you want it to behave like a Python programming teacher.

Instead of writing the same system prompt every time, you can create a custom model with predefined instructions.

# Create a Modelfile
create a Modelfile.txt or you can create a Modelfile only 
you just have to change during running command:
ollama create python-teacher -f Modelfile.txt or Modelfile as per creation


this is the content of modelfile:

FROM qwen2.5:3b

SYSTEM """
You are a Python programming teacher.
Explain concepts in simple language.
Always provide practical examples.
Avoid unnecessary theory.
"""

PARAMETER temperature 0.3

# Structured Output – Getting JSON Responses
Normally, LLMs return responses in natural language.

For example:
Python is a programming language used for AI,
web development and automation.

But sometimes applications need structured data.
For example, suppose you want an LLM to identify information from a sentence and return:

{
  "name": "Abhishek",

  "technology": "Python",

  "field": "AI"
}
This is called structured output.

# Tool Calling – Allowing LLMs to Use Python Functions
With the help of tool calling, an LLM can request the execution of external functions.
For example, an LLM normally cannot directly calculate the current result of a Python function unless your application gives it that capability.

You can provide a calculator function as a tool.
if you ask to llm what is 25*85 then LLM tries to calculate the answer itself.so you can use python function and llm will use the function to calculate this.
