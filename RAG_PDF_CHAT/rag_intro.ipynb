# Retrieval Augmented Generation (RAG)
It is a technique that allows an LLM (Large Language Model) to answer questions using external data that was not necessarily included in its training.

# RAG Architecture
RAG architecture consists of two main phases:
1. Indexing Pipeline – Preparing and storing documents.
    - Data Sources:
    PDF, DOCX, TXT, Websites, Databases
    - Document Loading:
    Extract text from documents
    - Text Chunking:
    Divide text into smaller chunks
    - Embedding Model:
    Convert chunks into numerical vectors
    - Vector Database:
    Store vectors, text chunks and metadata

2. Query Pipeline – Retrieving information and generating answers.

    - User Query
    "What is the company's leave policy?"
    - Query Embedding
    Convert question into a vector
    - Similarity Search
    Compare query vector with stored vectors
    - Retrieval
    Fetch the most relevant text chunks
    - Prompt Construction
    Combine retrieved context + user question
    - LLM
    Generate an answer using the provided context
    -Final Response
    Display the answer to the user
