# Convolutional Neural Network:

CNN is a type of Deep Learning model mainly used for image processing and computer vision.


CNN Architecture:
CNN architecture is the sequence of layers through which an image passes to identify and classify objects or patterns.


Main layers of CNN
Layer	Purpose
1. Input Layer	Takes the image as input.
2. Convolution Layer	Extracts features using filters/kernels.
3. ReLU Layer	Removes negative values and adds non-linearity.
4. Pooling Layer	Reduces feature map dimensions.
5. Flatten Layer	Converts 2D feature maps into 1D data.
6. Fully Connected Layer	Learns patterns for classification.
7. Output Layer	Produces the final prediction.


Input: CNN receives an image of a handwritten digit.

Convolution: Detects edges and patterns.

ReLU: Applies activation.

Pooling: Reduces image feature dimensions.

Flatten: Converts extracted features into a vector.

Fully Connected: Learns which digit the features represent.

Output: Predicts the digit, for example, 7.


feature extraction:
A filter (also called a kernel) is a small matrix of numbers that slides over an image to detect specific patterns

Why do we need filters?
Imagine giving a CNN a picture of a cat.

The CNN needs to identify:
- Edges of the cat.
- Shape of its ears.
- Texture of its fur.
- Eyes and nose.
- Overall patterns.
A filter helps detect these features from the image.

Different filters can detect different patterns.


Padding: Padding is the process of adding extra pixels (usually zeros) around the borders of an image before applying a filter/kernel.
It is used during convolution to:
1. Prevent image dimensions from shrinking too quickly.
2. Preserve information near image boundaries.
3. Control the output size of a feature map.


Stride: stride is the no of pixel that the kernal  moves of the input image.


Pooling: Pooling is a downsampling technique used in CNN to reduce the width and height of feature maps while retaining important information.

Pooling is used to:
1. Reduce the dimensions of feature maps.
2. Reduce computational cost.
3. Reduce the number of parameters in subsequent layers.
4. Help make feature detection less sensitive to small shifts in an image


Types of Pooling:

Max Pooling: Selects the maximum (largest) value from each region .max(1,3,5,6)=6

Min Pooling: takes minimum value from each origin.

Average Pooling
Calculates the average of all values in a region.(1+3+5+6) / 4=3.75

Global avg Pooling: avg of entire feature map 

Filters: Images contain thousands of pixel values. CNN uses filters to identify useful patterns instead of treating every pixel independently.

Early layers: edges and lines.

Middle layers: textures and shapes.

Deeper layers: object parts and complex patterns

Receptive Field: Receptive field is the region of the original input image that a particular neuron in a CNN can see or use to calculate its output.
