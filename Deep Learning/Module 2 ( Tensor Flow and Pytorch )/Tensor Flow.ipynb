# TensorFlow :
TensorFlow is an open-source machine learning and deep learning framework developed by Google.
TensorFlow is a Python-based framework that helps us build, train, and deploy Deep Learning models.
TensorFlow provides the tools and functionality to ML/DL tasks.
Example applications:
- Image classification
- Object detection
- Speech recognition
- Natural Language Processing (NLP)
- Time-series forecasting
- Face recognition
- Recommendation systems
Without TensorFlow, you would need to manually implement backpropagation and gradient calculations.
With TensorFlow, you can use an optimizer such as Adam to update model weights automatically.

What is a Tensor?
Tensor = Data structure
Flow = Flow of data through mathematical operations
A tensor is a multidimensional array used to represent data.
Think of it as an advanced version of a NumPy array.

0-D Tensor – Scalar -> one single value eg:5

1D Tensor – Vector
A collection of values [1,20,33]

2D Tensor – Matrix
Rows and columns.

3D Tensor – 3D array
A collection of matrices. Commonly used to represent image data with height, width and channels.

# Neural Network
A neural network is a computational model inspired by the structure of the human brain.

It consists of interconnected artificial neurons organized into layers.

Three main layers:
1. Input Layer – Receives data.
2. Hidden Layers – Learn patterns from data.
3. Output Layer – Produces predictions

Weights and Bias: 
These are learnable parameters of a neural network.
Weights: Determine how strongly one neuron influences another.
Bias: An additional adjustable value that helps the model shift its output.

The basic neuron equation:
y = wx+b

Where:
- \(x\) = Input
- \(w\) = Weight
- \(b\) = Bias
- \(y\) = Output before activation
The model learns suitable weights and biases during training

# Activation Function
An activation function introduces non-linearity into a neural network.
Without activation functions, even multiple layers would behave like a single linear transformation.

Important activation functions:

Function	Common usage
ReLU	    Hidden layers
Sigmoid	    Binary classification output
Softmax	    Multiclass classification output
Tanh	    Certain neural network architectures
Linear	    Regression output

# Forward Propagation
Forward propagation means passing input data through the neural network to generate predictions.
Example:
Input: House size = 1000 sq ft
Model prediction: ₹40 lakh
Actual price: ₹45 lakh

The model produces a prediction, which is then compared with the actual answer.

# Loss Function
A loss function measures the difference between the model's prediction and the actual value.
Example:
Actual price = 45
Predicted price = 40
Error = 5

Common loss functions:
Loss function	                Usage
Mean Squared Error (MSE)	    Regression
Mean Absolute Error (MAE)	    Regression
Binary Crossentropy	            Binary classification
Categorical Crossentropy	    Multiclass classification
Goal: Minimize the loss during training

Backpropagation
Backpropagation is the process of calculating how much each model parameter contributed to the prediction error.
In simple words:
The model makes a mistake, calculates how its weights contributed to that mistake, and determines how to adjust them


# Optimizer
An optimizer uses the calculated gradients to update the model's weights and biases.
Important optimizers:
- SGD (Stochastic Gradient Descent)
- Adam
- RMSprop
- Adagrad
Important difference:
- Loss function: Measures the error.
- Backpropagation: Calculates gradients.
- Optimizer: Updates weights using gradients.

```python
# tf.constant() and tf.variable() :
""" tf.constant creates an immutable tensor whose value doesn't change,
whereas tf.Variable creates a mutable tensor whose value can be updated.
Neural network weights and biases are typically stored as tf.Variable because they are updated during training."""

cns=tf.constant([10,20])
# cns[1]=15 error: object does not support item assignment
display(cns)
vrb=tf.Variable([10,20])
display(vrb)
vrb[1].assign(15) #  normal indexing doesnt work we have to assign through assing function if we want to make changes.
display(vrb)
```

```python
# generating random tensor : 
# from_seed(42) this tells that the number is being generated will same if you use same feed for other , but the initilization is not depend on you.
# so in true sense its not random. but you can understand what is going on here.
random1=tf.random.Generator.from_seed(42)
random1=random1.normal(shape=(3,2)) # normal= normal distribution
random2=tf.random.Generator.from_seed(42)
random2=random2.normal(shape=(3,2))
# to check are they equal?
random1,random2,random1==random2
```

```python
# shuffling the tensor:
print(random1)
sf=tf.random.shuffle(random1)
print(sf)
# you can make shuffling fix by passing seed
sf_Seed=tf.random.shuffle(random1,seed=42)
print(sf_Seed)
```

```python
# Converting the numpy array to tensor:
import numpy as np
numpy_A=np.arange(1,25,dtype=np.int32)
tf_A=tf.Variable(numpy_A.reshape(2,6,2)) # shape multiplication must be equal to np limit ( 1->25=24 so 2*3*4=must be 24)
print(numpy_A)
print(tf_A)
```

```python
rank4_tf.assign(
    tf.tensor_scatter_nd_update(
        rank4_tf,
        indices=[[0,0,1,2]],
        updates=[1]
    )
)
print(rank4_tf)
```
