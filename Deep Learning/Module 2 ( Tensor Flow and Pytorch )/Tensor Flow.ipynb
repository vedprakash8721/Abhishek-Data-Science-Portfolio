```python
import tensorflow as tf
from IPython.display import display
display(tf.__version__)
import warnings
warnings.filterwarnings('ignore')
```

```python
# Tensor: it is multidimensional array 
scaler=tf.constant(7)
print(scaler.ndim)
display(scaler)

vector=tf.constant([10,5])
print(vector.ndim)
display(vector)

metrix=tf.constant([[10,5],[7,6]])
print(metrix.ndim)
display(metrix)
```

```python
# tf.constant() and tf.variable() :
""" tf.constant creates an immutable tensor whose value doesn't change,
whereas tf.Variable creates a mutable tensor whose value can be updated.
Neural network weights and biases are typically stored as tf.Variable because they are updated during training."""

cns=tf.constant([10,20])
# cns[1]=15 error: object does not support item assignment
display(cns)
vrb=tf.Variable([10,20])
display(vrb)
vrb[1].assign(15) #  normal indexing doesnt work we have to assign through assing functio 
display(vrb)
```
