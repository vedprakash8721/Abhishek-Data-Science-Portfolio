# What is Agentic AI?
Agentic AI refers to AI systems that can pursue goals with some degree of autonomy by reasoning, planning, making decisions, using tools, observing outcomes and taking subsequent actions.

Autonomy: Performs work without requiring instructions for every individual step

An AI agent is a system that receives information about its environment, processes that information, decides what to do and takes actions to achieve a goal.

Think of an agent as a digital worker.

A digital worker has:
- A goal: What needs to be accomplished?
- A brain: How should it think and decide?
- Tools: What resources can it use?
- Memory: What information should it retain?
- Actions: What can it do?
- Feedback: Did the action produce the expected result?
For example, consider an AI research agent.

How does Agentic AI work?
An agent generally works through a repeated process called the Agent Execution Loop.

The basic execution loop contains five stages:
1. Perception / Input
2. Reasoning
3. Planning
4. Action
5. Observation and Feedback

Traditional AI simply recognizes what is in front of it, like telling if a picture is a cat or a dog.
( LLM ) Standard AI chat models just answer questions with text, like giving you interview tips when you ask. 
But Agentic AI goes a step further by acting on its own to plan, practice, and guide you through an entire goal from start to finish.

# Types of AI Agents
Simple reflex agent: Acts based on current conditions and predefined rules	If temperature is high, turn on cooling

Model-based reflex agent: Maintains an internal representation of the environment	Robot tracking obstacles

Goal-based agent	Selects actions to achieve a goal	Navigation system finding a destination

Utility-based agent	Selects actions based on a utility or preference measure	Choosing a route balancing time and cost

Learning agent	Improves its behaviour using experience or feedback	Recommendation system adapting to user behaviour

# Topic 4: Agent Architecture
Agent architecture means how an AI agent is structured and how its different components work together to complete a task.

One important concept is the ( Reason → Act → Observe ) loop.
- Reason: Decide what needs to be done.
- Act: Use a tool or perform an action.
- Observe: Check the result.
- Repeat: If the goal isn't achieved, decide the next action.

# Working of Agentic AI:
OBSERVE – How an AI Agent receives information
agenctic ai works as Observe → Think → Act → Observe cycle.

Observation is the process through which an agent receives information about the user request, its environment, or the result of an action.

# Types of observations
In practical agent systems, information commonly comes from three sources.

A. User input
Information directly provided by the user.

Examples:
- Calculate 50 × 20.

B. Environment observation:  
Information obtained from external systems or the environment.

Examples:
Source	            Observation
Website	            Search results
Database	        Customer records

The environment provides information that may not be present in the user's original message.

C. Tool output
Information returned after the agent executes a tool.

For example:
def calculator(a, b):
    return a * b

result = calculator(58, 79)

print(result)

The value 4582 becomes an observation for the agent.

# How does an LLM receive observations?
An LLM does not automatically know what happened in your Python application.

Your application must send the relevant information to the model through its input, usually messages.

# What is Context?
You will frequently hear the term context when working with LLMs and agents.

Context is the information provided to an LLM that helps it understand the current task and generate an appropriate response
It can include:
- User instructions.
- Previous conversation messages.
- Tool calls.
- Tool results.
- System instructions.
- Relevant retrieved information.




# THINK – How an AI Agent Makes Decisions
The Think step is where the agent uses its LLM and application logic to decide what action should happen next.

For example:

User: What is 58 multiplied by 79?

The agent receives the request and needs to decide:
- Can I answer directly?
- Do I need a calculator?
- Which tool should I use?
- What inputs should I provide?
It decides to use the calculator.




# Reasoning vs Decision-making vs Planning
A. Reasoning
Reasoning means analyzing available information to reach a conclusion.

Example:

User: "What is 58 multiplied by 79?"

The agent identifies:
- The user wants a multiplication.
- Two numbers are provided.
- A calculator is suitable.

B. Decision-making
Decision-making means selecting one action from the available choices.
for this multiplication agent selects the calculator

C. Planning
Planning means deciding the sequence of actions required to achieve a goal.

# GOALS – How an AI Agent Understands What It Needs to Achieve
What is a Goal in Agentic AI?
A goal is the desired outcome that an agent needs to achieve.

For example:

User: "Find three AI Engineer courses under ₹5,000 and compare them."

The goal is:
Find three suitable AI Engineer courses within the specified budget and compare them.

Notice that the goal describes the final outcome, not just one action.

The agent needs to perform several actions to achieve it

# PLANNING
Planning is the process of deciding which actions are required and in what order to achieve a goal.
Let's take an example.
User: Find three laptops under ₹60,000, compare them and recommend the best one.

Types of planning

A. Sequential planning
Tasks execute one after another.

Example:
Search laptops
      ↓
Extract prices
      ↓
Filter by budget
      ↓
Compare laptops
      ↓
Recommend

B. Parallel planning
Independent tasks can execute at the same time.

Example:



C. Dynamic planning
The agent changes its plan based on new observations.

Example:
1. Search for laptops.
2. Find only two laptops under ₹60,000.
3. Decide to broaden the search.
4. Search again.
5. Find a third suitable laptop.
6. Continue with comparison.

# What is the Act step? 
Act is the stage where an agent performs an action to make progress towards its goal.

An action could be:
- Calling a Python function.
- Searching the internet.
- Reading a file.
- Querying a database.
- Calling an API.
- Creating or updating a record

# The Complete Agent Loop 
An agent doesn't necessarily perform just one action. It can execute a tool, observe the result, decide what to do next, and repeat until the task is complete





# What are tools in AI agents? 
A tool is an external function or capability that an AI agent can use to perform an action.

An LLM is good at understanding language, generating text, summarizing, and reasoning. But it cannot automatically interact with your computer, search the internet, or access a database.

Tools give it those capabilities.
LLM decides what needs to be done; the tool performs the actual action.

# Types of tools used in AI agents
In real-world AI applications, tools can perform many different tasks.



Imagine asking your agent:

"Find today's weather in Bengaluru, calculate the temperature difference from yesterday, and save the result in a file."

It might use three tools:
1. Weather API → Get today's and yesterday's temperatures.
2. Calculator → Calculate the difference.
3. File tool → Save the result.

# Tool schemas, tool selection, and argument extraction 
A tool schema is a structured description of a tool that tells the LLM:
1. What is the tool's name?
2. What does it do?
3. What inputs does it require?
4. What data types are expected?

# Agent Workflows & Advanced Concepts

Agent vs Workflow : 
What is a workflow?
A workflow is a predefined sequence of steps used to complete a task.

For example, suppose you want to summarize a PDF:
1. Read the PDF.
2. Extract the text.
3. Split the text into chunks.
4. Summarize the text.
5. Save the summary.

What is an AI Agent?
An AI agent can decide what action to take based on the user's request, available tools, and results from previous actions.

For example:

User: "Find the latest AI news and summarize the top three stories."

The agent needs to:
1. Understand the request.
2. Decide to use web search.
3. Search for recent AI news.
4. Examine the results.
5. Decide what information is relevant.
6. Generate a summary.

# Lesson 2 – Agent Orchestration

What is orchestration?
Orchestration means coordinating different components, tools, and steps so they work together to complete a task.

Imagine your Jarvis assistant has these tools:
- Calculator
- Web search
- Date and time
- File reader
- Email sender
Orchestration decides:
- Which tool should execute?
- In what order?
- Should multiple tools execute together?
- What happens if a tool fails?
- When should the agent stop?
